```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
plt.rcParams.update(
        {
            "figure.figsize": (9.0, 5.5),
            "axes.grid": False,
            "grid.alpha": 0.3,
            "axes.labelsize": 20,
            "axes.titlesize": 20,
            "legend.frameon": False,
            "lines.linewidth": 2.5,
            "text.usetex": False,
            "font.family": "serif",
            "font.serif": "Computer Modern Roman",
            "xtick.labelsize": 16,
            "ytick.labelsize": 16,
            "legend.fontsize": 14,
        }
    )
```

```python
z = np.load("bp_accuracy_results.npz")
tlist = z["t_values"]
passive = z["acc_diffusion"]
active = z["acc_active"]
passive_equiv = z["acc_diffusion_equiv"]
```

```python
fig, ax = plt.subplots(figsize=(8.0, 5.0))
color1 = "blue"
color2 = "red"
color3 = "purple"
ax.semilogx(tlist, passive[10,:], label="Passive", alpha=0.5, color=color1)
ax.scatter(tlist, passive[10,:],color=color1)
ax.semilogx(tlist, active[10,:], label="Active", alpha=0.5, color=color2)
ax.scatter(tlist, active[10,:],color=color2)
ax.semilogx(tlist, passive_equiv[10,:], label="Equivalent Passive", alpha=0.5, color=color3)
ax.scatter(tlist, passive_equiv[10,:],color=color3)
ax.set_xlabel("Time")
ax.set_ylabel("Accuracy")
ax.legend()
plt.tight_layout()
plt.savefig("Root_bp_accuracy_results.pdf")
```

```python
fig, ax = plt.subplots(figsize=(8.0, 5.0))
cmap = plt.get_cmap("Reds")
level_count = 11
colors = cmap(np.linspace(0.15, 0.8, level_count))
for level in range(level_count):
    ax.semilogx(tlist, active[level,:], color=colors[level], label=f"Level {level}")
ax.set_xlabel("Time")
ax.set_ylabel("Accuracy")
ax.legend()
plt.tight_layout()
plt.savefig("Bp_accuracy_results_active.pdf")
```
